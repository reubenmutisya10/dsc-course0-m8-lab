### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
injury_cols = [
    "Total.Fatal.Injuries",
    "Total.Serious.Injuries",
    "Total.Minor.Injuries",
    "Total.Uninjured"
]

accidents[injury_cols] = accidents[injury_cols].fillna(0)

# Estimate total passengers
accidents["Total.Passengers"] = (
    accidents["Total.Fatal.Injuries"]
    + accidents["Total.Serious.Injuries"]
    + accidents["Total.Minor.Injuries"]
    + accidents["Total.Uninjured"]
)

# Total serious or fatal injuries
accidents["Serious.Fatal"] = (
    accidents["Total.Fatal.Injuries"]
    + accidents["Total.Serious.Injuries"]
)

# Injury rate
accidents["Serious.Fatal.Rate"] = (
    accidents["Serious.Fatal"] /
    accidents["Total.Passengers"].replace(0, pd.NA)
)

accidents[[
    "Total.Passengers",
    "Serious.Fatal",
    "Serious.Fatal.Rate"
]].head()
```

**Aircraft.Damage**
- identify and execute any cleaning tasks
- construct a derived column tracking whether an aircraft was destroyed or not.

```python
accidents["Aircraft.damage"].value_counts(dropna=False)
```

```python
accidents["Destroyed"] = (
    accidents["Aircraft.damage"]
    .fillna("Unknown")
    .eq("Destroyed")
)

accidents[["Aircraft.damage", "Destroyed"]].head()
```

### Inspect Model column
- Get rid of any NaNs.
- Inspect the column and counts for each model/make. Are model labels unique to each make?
- If not, create a derived column that is a unique identifier for a given plane type.

```python
# Remove rows where Make or Model is missing
accidents = accidents.dropna(subset=["Make", "Model"]).copy()

# Clean extra spaces and standardize capitalization
accidents["Make"] = accidents["Make"].str.strip().str.title()
accidents["Model"] = accidents["Model"].str.strip().str.upper()

# Inspect common models
accidents[["Make", "Model"]].value_counts().head(20)
```

```python
# Combine Make and Model to identify each aircraft type
accidents["Aircraft.Type"] = (
    accidents["Make"] + " " + accidents["Model"]
)

accidents[["Make", "Model", "Aircraft.Type"]].head()
```

### Cleaning other columns
- there are other columns containing data that might be related to the outcome of an accident. We list a few here:
- Engine.Type
- Weather.Condition
- Number.of.Engines
- Purpose.of.flight
- Broad.phase.of.flight

Inspect and identify potential cleaning tasks in each of the above columns. Execute those cleaning tasks. 

**Note**: You do not necessarily need to impute or drop NaNs here.

```python
other_columns = [
    "Engine.Type",
    "Weather.Condition",
    "Number.of.Engines",
    "Purpose.of.flight",
    "Broad.phase.of.flight"
]

# Inspect missing values
accidents[other_columns].isna().sum()
```

```python
text_columns = [
    "Engine.Type",
    "Weather.Condition",
    "Purpose.of.flight",
    "Broad.phase.of.flight"
]

for column in text_columns:
    accidents[column] = (
        accidents[column]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
    )

# Convert number of engines to numeric
accidents["Number.of.Engines"] = pd.to_numeric(
    accidents["Number.of.Engines"],
    errors="coerce"
)

accidents[other_columns].head()
```

### Column Removal
- inspect the dataframe and drop any columns that have too many NaNs

```python
# Percentage of missing values in each column
missing_percent = accidents.isnull().mean() * 100

missing_percent.sort_values(ascending=False)
```

```python
# Drop columns with more than 70% missing values
cols_to_drop = missing_percent[missing_percent > 70].index

accidents = accidents.drop(columns=cols_to_drop)

print("Dropped columns:")
print(cols_to_drop.tolist())

print("\nRemaining shape:", accidents.shape)
```
